# 07a · Start the preprocessing order check

Run this one cell with the **Python 3.12 (whiteboard-ocr)** kernel. It starts `scripts/preprocess_order_check.py` in the background (1-2 h, resumable) and returns at once; then run notebook 07.

```python
# Starts scripts/preprocess_order_check.py in the background (it keeps running if this notebook is closed).
# Progress: whiteboard-ocr/outputs/preprocess_order/run.log ; result: summary.txt and decision.json there.
import subprocess, sys
from pathlib import Path
WB = Path.cwd().resolve()
WB = WB if (WB / "preprocessing").is_dir() else WB.parent
log_dir = WB / "outputs" / "preprocess_order"; log_dir.mkdir(parents=True, exist_ok=True)
log = open(log_dir / "run.log", "a")
p = subprocess.Popen([sys.executable, "-u", str(WB / "scripts" / "preprocess_order_check.py")],
                     cwd=str(WB.parent), stdout=log, stderr=subprocess.STDOUT, start_new_session=True)
(log_dir / "run.pid").write_text(str(p.pid))
print("started, pid", p.pid, "-> log:", log_dir / "run.log")
```
